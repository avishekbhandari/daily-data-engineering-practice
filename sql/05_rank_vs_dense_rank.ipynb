
Topic: Ranking Records Using ROW_NUMBER(), RANK(), and DENSE_RANK()

Purpose:
This script demonstrates how ROW_NUMBER(), RANK(), and DENSE_RANK()
behave when ranking records within each category.

Pattern:
This approach is commonly used in data engineering and analytics for top-N analysis,
product ranking, customer ranking, provider ranking, and leaderboard-style reporting.


```python
%sql
DROP TABLE IF EXISTS product_sales;

CREATE TABLE product_sales (
    product_id INT PRIMARY KEY,
    product_name VARCHAR(100) NOT NULL,
    category VARCHAR(50) NOT NULL,
    total_sales DECIMAL(10, 2) NOT NULL
);

INSERT INTO product_sales (product_id, product_name, category, total_sales) VALUES
(1, 'Laptop', 'Electronics', 5000.00),
(2, 'Monitor', 'Electronics', 3200.00),
(3, 'Tablet', 'Electronics', 3200.00),
(4, 'Mouse', 'Electronics', 900.00),
(5, 'Keyboard', 'Electronics', 900.00),
(6, 'Desk', 'Furniture', 4200.00),
(7, 'Office Chair', 'Furniture', 4200.00),
(8, 'Bookshelf', 'Furniture', 2600.00),
(9, 'Desk Lamp', 'Furniture', 1500.00),
(10, 'Notebook', 'Stationery', 800.00),
(11, 'Pen Set', 'Stationery', 800.00),
(12, 'Marker Pack', 'Stationery', 500.00),
(13, 'Stapler', 'Stationery', 300.00);

SELECT
    product_id,
    product_name,
    category,
    total_sales,
-- product_id is added to ORDER BY as a tie-breaker to ensure deterministic (consistent) row numbers when total_sales are identical
    ROW_NUMBER() OVER (
        PARTITION BY category
        ORDER BY total_sales DESC, product_id
    ) AS row_number_rank,

    RANK() OVER (
        PARTITION BY category
        ORDER BY total_sales DESC
    ) AS rank_value,

    DENSE_RANK() OVER (
        PARTITION BY category
        ORDER BY total_sales DESC
    ) AS dense_rank_value

FROM product_sales
ORDER BY
    category,
    total_sales DESC,
    product_id;
```
